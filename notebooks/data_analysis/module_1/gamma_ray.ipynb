# Data background and initial problem setup

The file gamma-ray.csv contains a small quantity of data collected from the Compton Gamma Ray Observatory, a satellite launched by NASA in 1991 (http://cossc.gsfc.nasa.gov/). For each of 100 sequential time intervals of variable lengths (given in seconds), the number of gamma rays originating in a particular area of the sky was recorded. You would like to check the assumption that the emission rate is constant.

OK, data looks like a number of durations and then associated counts. If we are going to model this data, natural choice is 
the Poisson model which estimates the number of successes observed during a continuous window when the rate of success is known and constant

# Poisson model explainer


The Poisson model provides the probability of observing $k$ successes during a certain interval given that a success happens at a constant rate $r$. It is a discrete distribution and its Probability Mass Function (PMF) takes only one parameter $\lambda$ which is the expected number of events in the window that you are counting over, $t$. The relationship between $\lambda$, $r$ and $t$ is $\lambda = rt$. The PMF of the Poisson distribution is given by:

$$
P(X=k) = \frac{\lambda^k e^{-\lambda}}{k!}
$$

It should be noted that some textbooks define $\lambda$ as the rate itself and write Poi($\lambda t$) instead of Poi($\lambda$).

The Poisson distribution requires the following assumptions:  
 - events arrive in continues time
 - events are independent
 - counts are dependent only on the length of the interval, not on where the interval starts or ends
 - simultaneous events are impossible

 The core Poisson distribution properties are:  

 | Property | Value |
 |---|---|
 | Mean | $\lambda$ |
 | Variance | $\lambda$ |

 - This means that the Poisson distribution is equidispersive (equal mean and variance). This is an assumption that is most often broken by real data. 

 - The Poisson distribution is additive meaning if X ~Poi($\lambda_1$) and Y ~Poi($\lambda_2$), then X + Y ~(Poi($\lambda_1$) + Poi($\lambda_2$))
